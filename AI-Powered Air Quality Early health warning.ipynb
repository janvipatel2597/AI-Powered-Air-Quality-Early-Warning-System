```python
import pandas as pd
import numpy as np

profiles   = pd.read_csv('/content/drive/MyDrive/raw_user_profiles.csv')
activities = pd.read_csv('/content/drive/MyDrive/raw_user_activities.csv')

print("Profiles shape:", profiles.shape)
print("Activities shape:", activities.shape)
```

```python
profiles['creation_date']  = pd.to_datetime(profiles['creation_date'])
activities['timestamp']    = pd.to_datetime(activities['timestamp'])

print("Profiles creation_date dtype:", profiles['creation_date'].dtype)
print("Activities timestamp dtype:", activities['timestamp'].dtype)
```

```python
profiles['home_city']    = profiles['home_city'].replace('Unknown', np.nan)
profiles['home_region']  = profiles['home_region'].replace(['Unknown', 'Everywhere'], np.nan)
profiles['home_country'] = profiles['home_country'].replace('Unknown', np.nan)

activities['post_city']   = activities['post_city'].replace('Unknown', np.nan)
activities['post_region'] = activities['post_region'].replace('Everywhere', np.nan)

print("Unknowns remaining in profiles city:", (profiles['home_city'] == 'Unknown').sum())
print("Unknowns remaining in activities city:", (activities['post_city'] == 'Unknown').sum())
```

```python
bool_cols_profiles = [
    'is_private', 'is_verified', 'profile_picture',
    'profile_banner', 'has_bio', 'has_website',
    'has_location', 'is_fake'
]
for col in bool_cols_profiles:
    profiles[col] = profiles[col].astype(int)

bool_cols_activities = ['is_weekend', 'has_media', 'contains_url', 'is_fake']
for col in bool_cols_activities:
    activities[col] = activities[col].astype(int)

print("Boolean conversion done ✅")
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

profiles['account_type_encoded'] = le.fit_transform(profiles['account_type'])
profiles['language_encoded']     = le.fit_transform(profiles['language_preference'])

activities['device_encoded']       = le.fit_transform(activities['device'])
activities['platform_encoded']     = le.fit_transform(activities['platform'])
activities['media_type_encoded']   = le.fit_transform(activities['media_type'])
activities['content_type_encoded'] = le.fit_transform(activities['content_type'])
activities['language_encoded']     = le.fit_transform(activities['language'])

print("Encoding done ✅")
```

```python
profiles_clean = profiles.drop(columns=[
    'username', 'full_name', 'first_name', 'last_name',
    'email', 'home_country', 'home_region', 'home_city',
    'account_type', 'language_preference'
])

activities_clean = activities.drop(columns=[
    'activity_id', 'content',
    'post_country', 'post_region', 'post_city',
    'device', 'platform', 'media_type', 'content_type', 'language'
])

print("Profiles columns:", list(profiles_clean.columns))
print("Activities columns:", list(activities_clean.columns))
```

```python
print("===== PROFILES =====")
print("Shape:", profiles_clean.shape)
print("Missing values:\n", profiles_clean.isnull().sum())

print("\n===== ACTIVITIES =====")
print("Shape:", activities_clean.shape)
print("Missing values:\n", activities_clean.isnull().sum())
```

```python
profiles_clean.to_csv('/content/drive/MyDrive/clean_user_profiles.csv', index=False)
activities_clean.to_csv('/content/drive/MyDrive/clean_user_activities.csv', index=False)

print("✅ clean_user_profiles.csv saved to Google Drive")
print("✅ clean_user_activities.csv saved to Google Drive")
```
